```python
import pandas as pd
```

```python
# Leer el archivo Excel
file_name = "../mis_direcciones.xlsx"
excel_data = pd.read_excel(file_name)
```

```python
# Crear DataFrames únicos para cada entidad
provincias_df = excel_data[['Provincia']].drop_duplicates().reset_index(drop=True)
cantones_df = excel_data['Canton'].drop_duplicates().reset_index(drop=True)
parroquias_df = excel_data['Parroquia'].drop_duplicates().reset_index(drop=True)
```

```python
# Mostrar las tablas
print(f"Provincias: \n{provincias_df.head()}")
print(f"\nCantones: \n{cantones_df.head()}")
print(f"\nParroquias: \n{parroquias_df.head()}")
```
