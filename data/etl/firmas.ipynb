```python
# Leer el archivo de texto con Polars
file_path = '../processing/PSR-1100f_SN22A44H3_00037.txt'
```

```python
# Función para procesar líneas del archivo y extraer solo Reflect (Y)
def extraer_reflect(line):
    try:
        return float(line.strip().split('\t')[1])  # Extrae Reflect. % (segunda columna)
    except (ValueError, IndexError):
        return None  # Ignorar líneas no válidas

# Leer el archivo y procesar solo la columna Reflect (Y)
with open(file_path, 'r') as file:
    reflect_data = [extraer_reflect(line) for line in file if "\t" in line]

# Filtrar valores válidos (eliminar None)
reflect_data = [reflect for reflect in reflect_data if reflect is not None]

# Aplanar los datos
y_list = np.array(reflect_data, dtype=float)

# Mostrar la lista de Reflect. % (Y)
y_list
# mean = np.mean(y_list, axis=0, dtype=float)
# mean
```

```python
# Función para procesar líneas del archivo y extraer solo Reflect (Y)
def extraer_wvl(line):
    try:
        return float(line.strip().split('\t')[0])  # Extrae Reflect. % (segunda columna)
    except (ValueError, IndexError):
        return None  # Ignorar líneas no válidas

# Leer el archivo y procesar solo la columna Reflect (Y)
with open(file_path, 'r') as file:
    wvl_data = [extraer_wvl(line) for line in file if "\t" in line]

# Filtrar valores válidos (eliminar None)
wvl_data = [wvl for wvl in wvl_data if wvl is not None]

# Aplanar los datos
x_list = np.array(wvl_data, dtype=float)

# Mostrar la lista de Reflect. % (Y)
x_list
```
